```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate, FewShotPromptTemplate
```

```python
load_dotenv
Api_key = os.getenv
```

```python
examples = [
    {
        "message" : "My package hasnt arrived",
        "category" : "shipping"
     },

     {
         "message":"I was charged twice for my order",
         "category":"Billing"
     },
     {
         "message":"The app crashes when I try to log in.",
         "category":"Technical Support"
     },
     {
         "message":"I want to return the shoes I bought. ",
         "category":"Returns"
     },
     {
         "message":"why was I charges an extra fee",
         "category":"billing"
     },
     {
         "message":"The websites keeps showing an error",
         "category":"Technical support"
     },
     {
         "message":"I forgot my password.",
         "category":"Account"
     },
     {
         "message":"how do i change my account email ?",
         "category":"Account"
     }
    

]
```

```python
example_prompt = PromptTemplate.from_template(
    "message:{message} \nCategory: {category}"
)

prompt_template = FewShotPromptTemplate(
    examples = examples,
    example_prompt = example_prompt,
    suffix = "message:{input}",
    input_variables = ["input"],

)

prompt = prompt_template.invoke({"input":"i havent recived my funds yet."})
print(prompt.to_string())
```

```python
#connect llm
llm = ChatOpenAI(
    model="gpt-4.1-mini",
    temperature=0
)

llm_chain = prompt_template | llm
```

```python
response = llm_chain.invoke({
    "input": "Where is my order."
}

)

print(response.content)
```
